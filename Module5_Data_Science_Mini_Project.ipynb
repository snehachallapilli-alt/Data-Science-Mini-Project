```python

# Step 4: Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python


from google.colab import files
import pandas as pd
import io

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("Dataset loaded successfully!")
print("Dataset filename:", filename)
print("\nFirst 5 rows of the dataset:")
display(df.head())
```

```python

# Step 7: Understand the Dataset Structure

print("Number of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nDataset size:")
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

```python

# Step 8: Check for Missing Values

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values in the dataset:")
print(df.isnull().sum().sum())

print("\nMissing value check completed!")
```

```python

# Step 9: Statistical Summary of Student Scores

print("Statistical Summary of the Dataset:")
display(df.describe())

print("\nAverage Scores:")
print("Math Score:", df["math score"].mean())
print("Reading Score:", df["reading score"].mean())
print("Writing Score:", df["writing score"].mean())

print("\nHighest Scores:")
print("Math:", df["math score"].max())
print("Reading:", df["reading score"].max())
print("Writing:", df["writing score"].max())
```

```python

# Step 10: Average Scores by Gender

gender_scores = df.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean()

print("Average Scores by Gender:")
display(gender_scores.round(2))

print("\nAnalysis completed successfully!")
```

```python

# Step 11: Visualize Average Scores by Gender

gender_scores.plot(kind="bar", figsize=(9, 5))

plt.title("Average Student Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

```python

# Step 12: Average Scores by Parental Education

parent_education_scores = df.groupby(
    "parental level of education"
)[["math score", "reading score", "writing score"]].mean()

print("Average Scores by Parental Education:")
display(parent_education_scores.round(2))

# Visualize the results
parent_education_scores.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Average Student Scores by Parental Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Score")
plt.xticks(rotation=35, ha="right")
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

```python

# Step 13: Analyze Test Preparation Course

preparation_scores = df.groupby(
    "test preparation course"
)[["math score", "reading score", "writing score"]].mean()

print("Average Scores by Test Preparation Course:")
display(preparation_scores.round(2))

# Create a bar chart
preparation_scores.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Average Scores by Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

```python

# Step 14: Distribution of Math Scores

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="math score",
    bins=15,
    kde=True
)

plt.title("Distribution of Student Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

# Step 15: Compare Math, Reading, and Writing Scores

plt.figure(figsize=(9, 5))

score_data = df[["math score", "reading score", "writing score"]]

sns.boxplot(data=score_data)

plt.title("Comparison of Student Scores Across Subjects")
plt.xlabel("Subject")
plt.ylabel("Score")
plt.tight_layout()
plt.show()
```

```python

# Step 16: Correlation Between Student Scores

score_data = df[["math score", "reading score", "writing score"]]

correlation = score_data.corr()

print("Correlation Matrix:")
display(correlation.round(2))

# Create a correlation heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Student Scores")
plt.tight_layout()
plt.show()
```

```python

# Step 17: Analyze Student Scores by Lunch Type

lunch_scores = df.groupby(
    "lunch"
)[["math score", "reading score", "writing score"]].mean()

print("Average Scores by Lunch Type:")
display(lunch_scores.round(2))

# Visualize the comparison
lunch_scores.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Average Student Scores by Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

```python

# Step 18: Calculate Total and Average Scores

df["total score"] = (
    df["math score"]
    + df["reading score"]
    + df["writing score"]
)

df["average score"] = df["total score"] / 3

print("Dataset with Total and Average Scores:")
display(
    df[
        [
            "math score",
            "reading score",
            "writing score",
            "total score",
            "average score"
        ]
    ].head()
)

print("\nOverall Average Score:", round(df["average score"].mean(), 2))
print("Highest Total Score:", df["total score"].max())
print("Lowest Total Score:", df["total score"].min())
```

```python

# Step 19: Identify High-Performing Students

high_performers = df[df["average score"] >= 80]

print("Students with an Average Score of 80 or Above:")
display(
    high_performers[
        [
            "gender",
            "math score",
            "reading score",
            "writing score",
            "average score"
        ]
    ].head(10)
)

print("\nTotal High-Performing Students:", len(high_performers))
print("Percentage of High-Performing Students:",
      round((len(high_performers) / len(df)) * 100, 2), "%")
```

```python

# Step 20: Distribution of Average Student Scores

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="average score",
    bins=20,
    kde=True
)

plt.axvline(
    df["average score"].mean(),
    linestyle="--",
    label="Overall Mean"
)

plt.title("Distribution of Average Student Scores")
plt.xlabel("Average Score")
plt.ylabel("Number of Students")
plt.legend()
plt.tight_layout()
plt.show()
```

```python

# Step 21: Generate Key Project Findings

print("=" * 50)
print("STUDENT PERFORMANCE DATA ANALYSIS")
print("=" * 50)

print("\n1. Dataset Overview")
print("Total Students:", len(df))
print("Total Columns:", df.shape[1])

print("\n2. Average Subject Scores")
print("Math:", round(df["math score"].mean(), 2))
print("Reading:", round(df["reading score"].mean(), 2))
print("Writing:", round(df["writing score"].mean(), 2))

print("\n3. Highest Subject Scores")
print("Math:", df["math score"].max())
print("Reading:", df["reading score"].max())
print("Writing:", df["writing score"].max())

print("\n4. Test Preparation Comparison")
display(
    df.groupby("test preparation course")[
        ["math score", "reading score", "writing score"]
    ].mean().round(2)
)

print("\n5. Students with Average Score >= 80")
print(len(df[df["average score"] >= 80]))

print("\n6. Overall Average Student Score")
print(round(df["average score"].mean(), 2))

print("\nProject findings generated successfully!")
```
